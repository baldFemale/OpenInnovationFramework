# Evaluator: performance-based visibility selection strength

**Source simulation:** `Visibility_rank_selection(9).py`. All 100 senders are eligible for rule-based selection (`visibility_extent=1.0` in the script), but recipients receive a common pool of **exactly 10 sender-solution entries** every **10 search periods**. Thus, the effective visibility extent is **0.1**. Selection strength controls the fraction selected by true NK fitness rank (smaller rank = better performance). The remaining slots are filled from a common random ordering of sender identities fixed at each crowd reset, skipping already selected senders.

- `rank_selection_strength = 0.0`: the same 10 randomly designated senders throughout each run (the fixed-sender limited-visibility benchmark; not search-apart).
- `rank_selection_strength = 1.0`: the 10 highest-performing sender entries at each visibility event.
- Intermediate strengths `0.2, 0.4, 0.6, 0.8` prioritize 2, 4, 6, 8 entries, respectively, with the remaining slots filled from the fixed random ordering.

The common pool is shuffled once per visibility event and evaluated in the same order by all recipients. Selection determines pool membership, not evaluation order. Recipients adopt at most the first perceived improvement.

All seven outcomes refer to the recipient crowd. Lower breakthrough solution ranks indicate better discovery. Pairwise diversity measures normalized Hamming distance; adoption quality reflects true NK fitness of adopted solutions. Generative potential uses independent search continuations from sampled successful adoptions (`generative_horizon=30`, `generative_replications=30`, `generative_sample_size=5`); its precise definition resides in `Crowd.calculate_generative_potential`. The figures equally average outcomes across `K=9..17`.

The script uses **400 landscapes × 5 solver resets** (2,000 runs) per `(selection strength, K)` before saving aggregated means. Saved pickle files contain aggregate outcomes, not individual-run observations.

```python
from pathlib import Path
import pickle
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display

# Matched to the earlier evaluator notebook.
legend_properties = {'weight': 'bold'}
nus_blue = '#003D7C'
nus_orange = '#EF7C00'
morandi_blue = '#046586'
morandi_green = '#28A9A1'
morandi_yellow = '#C9A77C'
morandi_orange = '#F4A016'
morandi_pink = '#F6BBC6'
morandi_red = '#E71F19'
morandi_purple = '#B08BEB'
shallow_grey = '#D3D4D3'

K_colors = [morandi_blue, morandi_orange, morandi_green, morandi_purple,
            morandi_red, morandi_yellow, '#555555', '#B75F91']


def format_axis(ax):
    """Keep the formatting of the provided evaluator."""
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)
        spine.set_color('black')
    ax.grid(alpha=0.5)
    ax.tick_params(axis='both', labelsize=11)
    return ax


def safe_nanmean(values, axis=None):
    """Ignore structurally undefined adoption values, retaining all-NaN as NaN."""
    with warnings.catch_warnings():
        warnings.simplefilter('ignore', category=RuntimeWarning)
        return np.nanmean(values, axis=axis)
```

## 2. Load and validate the seven outcome families

The simulation saves one pickle per treatment and outcome with the exact filename pattern:

`visibility_rank_selection_fixed_random_strength_{strength}_recommendation_num_{number}_interval_{interval}_{DV}_across_K_size_{agent_num}`

Six scalar outcomes have shape `(rank selection strength, K)`, including the new `generative_potential`. The cumulative adoption-fitness trajectory has shape `(rank selection strength, K, visibility event)`. The loader validates **all six strengths**, **nine values of K**, and **30 visibility events**. Missing or invalid outcomes remain `NaN` in non-strict mode.

```python
scalar_dvs = [
    'breakthrough_rank',
    'breakthrough_fitness',
    'pairwise_diversity',
    'unique_solution_count',
    'adopted_solution_fitness',
    'generative_potential',
]
trajectory_dv = 'adopted_solution_fitness_across_time'
expected_events = search_iteration // visibility_interval


def result_file(strength, dv):
    filename = (
        f'visibility_rank_selection_fixed_random_strength_{strength}'
        f'_recommendation_num_{recommendation_num}_interval_{visibility_interval}'
        f'_{dv}_across_K_size_{agent_num}'
    )
    return result_folder / filename


def load_pickle(path):
    with open(path, 'rb') as infile:
        return pickle.load(infile)


def load_visibility_rank_results(strict=True):
    n_strength = len(rank_selection_strength_list)
    n_K = len(K_list)
    data = {dv: np.full((n_strength, n_K), np.nan) for dv in scalar_dvs}
    data[trajectory_dv] = np.full((n_strength, n_K, expected_events), np.nan)
    missing_files = []
    invalid_files = []

    for i, strength in enumerate(rank_selection_strength_list):
        for dv in scalar_dvs + [trajectory_dv]:
            path = result_file(strength, dv)
            if not path.is_file():
                missing_files.append(path.name)
                continue
            try:
                values = np.asarray(load_pickle(path), dtype=float)
            except (OSError, EOFError, pickle.UnpicklingError, TypeError, ValueError) as exc:
                invalid_files.append((path.name, str(exc)))
                continue
            shape = (n_K, expected_events) if dv == trajectory_dv else (n_K,)
            if values.shape != shape:
                invalid_files.append((path.name, f'got {values.shape}; expected {shape}'))
                continue
            data[dv][i] = values

    n_expected = n_strength * (len(scalar_dvs) + 1)  # Six scalar DVs plus one trajectory DV.
    print(f'Files loaded: {n_expected - len(missing_files) - len(invalid_files)} / {n_expected}')
    print('Array shapes:', {dv: value.shape for dv, value in data.items()})

    if missing_files:
        print(f'\nMissing files ({len(missing_files)}):')
        for filename in missing_files[:12]:
            print('  ', filename)
        if len(missing_files) > 12:
            print('  ...')
    if invalid_files:
        print(f'\nInvalid files ({len(invalid_files)}):')
        for name, issue in invalid_files[:12]:
            print('  ', name, ':', issue)

    if strict and (missing_files or invalid_files):
        raise FileNotFoundError(
            'Missing/invalid selection-strength results. Check result_folder, agent_num, '
            'recommendation_num, visibility_interval, and rank_selection_strength_list; '
            'or set strict_loading=False for incomplete runs. First expected file: '
            + str(result_file(rank_selection_strength_list[0], 'breakthrough_rank'))
        )
    if not any(np.isfinite(data[dv]).any() for dv in scalar_dvs):
        raise FileNotFoundError('No valid scalar pickle outputs found in ' + str(result_folder))
    return data


data = load_visibility_rank_results(strict=strict_loading)
```

```python
# =========================
# Data checks and long-form details
# =========================
checks = {
    'breakthrough_rank': (1, 2 ** N),
    'breakthrough_fitness': (0, 1),
    'pairwise_diversity': (0, 1),
    'unique_solution_count': (1, agent_num),
    'adopted_solution_fitness': (0, 1),
}
for dv, (minimum, maximum) in checks.items():
    finite = data[dv][np.isfinite(data[dv])]
    outside = np.count_nonzero((finite < minimum - 1e-8) | (finite > maximum + 1e-8))
    print(f'{dv:29s} finite={finite.size:3d} / {data[dv].size:3d}; outside expected range={outside}')

# The generative-potential scale depends on Crowd.calculate_generative_potential;
# check coverage without imposing an unsupported numeric range.
print('generative_potential'.ljust(29),
      f"finite={np.isfinite(data['generative_potential']).sum():3d} / {data['generative_potential'].size:3d}")

# The trajectory's final cumulative mean should match the scalar adoption-quality
# measure for each (rank_selection_strength, K), up to floating-point differences.
last_trajectory = data[trajectory_dv][:, :, -1]
both_defined = np.isfinite(last_trajectory) & np.isfinite(data['adopted_solution_fitness'])
if both_defined.any():
    gap = np.max(np.abs(last_trajectory[both_defined] - data['adopted_solution_fitness'][both_defined]))
    print('Max discrepancy: final trajectory vs. scalar adoption quality =', round(float(gap), 12))

rows = []
for i, strength in enumerate(rank_selection_strength_list):
    for j, K in enumerate(K_list):
        row = {'rank_selection_strength': strength, 'K': K}
        for dv in scalar_dvs:
            row[dv] = data[dv][i, j]
        rows.append(row)

detailed_results = pd.DataFrame(rows)
print('\nDetailed data preview (one observation per rank selection strength × K):')
# display(detailed_results.head(12).round(4))
```

## 6. Cumulative adoption-quality trajectories

Set `trajectory_rank_selection_strength` to any of the six simulated strengths. The horizontal axis gives visibility-event periods (`10, 20, ..., 300`). Each value is the **cumulative mean true fitness of adopted sender solutions through that event**, rather than the mean fitness of all recipient solutions or adoptions in only that period. Unlike a search-apart condition, the `0.0` fixed random-selection condition can have a valid adoption-fitness trajectory.
